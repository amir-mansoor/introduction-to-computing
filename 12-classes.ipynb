## Inheritance Example

```python
class Polygon:
    def __init__(self,no_of_sides): # Called a constructor
        self.n = no_of_sides
        self.sides = [0 for i in range(no_of_sides)]
        
    def set_sides(self, sides):
        self.sides = sides
        
    def show_sides(self):
        for i in self.sides:
            print(f"Side {i}")
```

```python
q = Polygon(3)
```

```python
q.show_sides()
```

```python
q.set_sides([10,20,30])
```

```python
q.show_sides()
```

```python
p = Polygon(4)
```

```python
q.n
```

```python
p.n
```

```python
p.show_sides()
```

```python

p.set_sides([100,200,300,400])
```

```python
p.show_sides()
```

## Triangle

```python
class Triangle(Polygon):
    def __init__(self):
        Polygon.__init__(self,3)
        
    def get_area(self):
        a,b,c = self.sides
        
        s = (a+b+c) / 2
        area = (s * (s-a) * (s-b) * (s-c)) ** 0.5 # just the formula
        return area 
```

```python
t = Triangle()
```

```python
t.show_sides()
```

```python
t.get_area()
```

```python
t.set_sides([3,4,5])
```

```python
t.show_sides()
```

```python
t.get_area()
```

## Rectangle

```python
class Rectangle(Polygon):
    def __init__(self):
        Polygon.__init__(self,4)
        
    def set_sides(self,sides):
        assert sides[0] == sides[2], "Opposite sides must be equal"
        assert sides[1] == sides[3], "Opposite sides must be equal"
        self.sides = sides 
        
    def get_area(self):
        area = self.sides[0] * self.sides[1]
        return area
```

```python
r = Rectangle()
```

```python
r.set_sides([10,20,10,20])
```

```python
r.get_area()
```
