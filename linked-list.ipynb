## Linking the nodes

```python
class ListNode:
    def __init__(self,data):
        self.data = data
        self.next = None
        
class LinkedList:
    def __init__(self):      # head of the list
        self._start  = None
        
    def insert(self,data):
        next_node = ListNode(data) # create a new node
        
        if self._start is None: # first time creating a node
            self._start = next_node # set the start and current to first node
            self._current = next_node
        else:
            self._current.next = next_node # link the new node
            self._current = next_node     # update current position / start is always same
```

```python
my_list = LinkedList()
```

```python
my_list.insert(25)
```

```python
my_list._start.data
```

```python
my_list.insert(35)
```

```python
my_list._start.next.data
```

## This is getting difficult to work with. let's simplify this by writing a function that pretty prints the list

```python
def __str__(self):
    out_str = '['
    _curr = self._start
    
    while _curr is not None:
        out_str += str(_curr.data) + ", "
        _curr = _curr.next
        
    return out_str.strip(", ") + ']'
```

```python
LinkedList.__str__ = __str__
```

```python
my_list = LinkedList()
```

```python
my_list.insert(5)
my_list.insert(7)
my_list.insert(8)
print(my_list)
```

### Callbacks
we want to call a function on all nodes

```python
def print_squares(node_x):
    print(f"{node_x.data} x {node_x.data} = {node_x.data * node_x.data}")
```

```python
print_squares(my_list._start)
```

```python
def call_func(self,fn):
    _curr = self._start
    while _curr is not None:
        fn(_curr)
        _curr = _curr.next


LinkedList.call_func = call_func
```

```python
print(my_list)
my_list.call_func(print_squares)
```
