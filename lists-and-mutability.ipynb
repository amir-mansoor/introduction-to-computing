```python
mammals = ["cat","kangroo", "horse"]
fish = ["tuna", "shark", "catfish"]
```

```python
animals = [mammals,fish]
animals
```

```python
mammals[0] = "mule"
print(f"mammals {mammals}")
print(f"animals {animals}")
```

## Let's see this tutor

```python
mammals_2 = mammals
mammals_2[0] = "dog"
print(f"mammals_2 {mammals_2}")
print(f"animals {animals}")
print(f"mammals {mammals}")
```

```python
## so it's the same lst with different names but can we make a copy of the list
```

```python
mammals_3 = mammals[:] # this create a copy 
```

```python
mammals_3[0] = "mouse"
```

```python
print(f"mammals_3 {mammals_3}")
print(f"mammals {mammals}")
print(f"animals {animals}")
    
```

```python
s1 = ["student 1", "ali", "khan"]
s2 = ["student 2", "zia", "farooq"]
s3 = ["student 3", "hadi", "zaman"]
```

```python
students = [s1,s2,s3]
```

```python
students
```

```python
students_copy = students[:]
```

```python
students_copy
```

```python
students_copy[0][0] = "student 1 changed"
print(students_copy)
print(students)
```

```python
# ok this is unexpected
```

```python
import copy
```

```python
students_3 = copy.deepcopy(students)
```

```python
students_3
```

```python
students_3[0][0] = "student 1 changed again"
print(f"students {students}")
print(f"students_3 {students_3}")
```

```python
def change_var(x):
    x = x+10
    
def change_lst(lst):
    lst.append(10)
```

```python
x = 25
change_var(x)
x
```

```python
lst = [1,2,3,4]
change_lst(lst)
lst
```
