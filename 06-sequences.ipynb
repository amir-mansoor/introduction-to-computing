### Let's do some intereseting stuff with lists 

```python
def make_tea(ing):
    for i in ing:
        print(f"Put {i} in")
        
ing = ['water','cinnamon', 'tea leaves','sugar','milk']
make_tea(ing)
```

```python
s = [1,2,8,4,6,8,9,15]
```

```python
def count(s,value):
    """count how many times value has occur"""
    counter = 0
    for i in s:
        if i == value:
            counter = counter + 1
            
    return counter
```

```python
count(s,20)
```
