```python
directory = 'res'
name = 'dummy.txt'
```

```python
import os
```

```python
filename = os.path.join(directory,name)
print(filename)
```

```python
help(os.path)
```

```python
f = open(filename, 'r') # handle
for line in f:
    print(line)
    
f.close()
```

```python
f = open(filename, 'r') # handle

for line in f:
    print(line)
    
f.close() # never forget this
```

### That has blank lines that shouldn't be there 

```python
f = open(filename,'r')

for line in f:
    print(line.strip())
    
f.close()
```

```python
# A better way is to use a context manager

with open(filename, 'r') as f:
    for line in f:
        print(line.strip())
        
# no need to call f.close() 
```

```python
with open(filename, 'r') as f:
    print(f.read())
    
```

```python
'a-b-c'.split("-")
```

```python
l = []
with open(filename, 'r') as f:
    l = f.read().split("\n")
    
l
```
