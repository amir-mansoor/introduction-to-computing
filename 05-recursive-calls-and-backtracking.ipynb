```python
def absolute_value(x):
    """Compute abs value of x"""
    if x >= 0:
        return x
    else:
        return -x 
```

```python
### Another way to write this
def absolute_value(x):
    if not x <= 0:
        return x
    else:
        return -x 
```

```python
def is_weekday(day,month,year):
    import datetime
    return datetime.datetime(year=year,month=month,day=day).weekday()
```

```python
if(is_weekday(27,4,2026)):
    print("It's a weekday")
else:
    print("Weekend!!")
```

```python
def sqrt(x,guess=1):
    if x < 0:
        return None
    if(good_enough(guess,x)):
        return guess
    else:
        new_guess = improve_guess(guess,x)
        return sqrt(x,new_guess)
```

```python
def good_enough(guess,x):
    if(abs(guess*guess - x) < 0.1):
        return True
    else:
        return False
    
def improve_guess(guess,x):
    if(guess < x):
        return guess + 0.1
    else:
        return guess - 0.1
```
