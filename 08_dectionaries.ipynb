### Iterating Over Dictionaries

```python
for a in d:
    print(a)
```

```python
for k in d:
    v = d[k]
    print(f"{k} = {v}")
```

```python
'package' in d
```

```python
'else' in d
```

```python
# d['else'] = "Hey i have added else"

if 'else' not in d: # you can move the not
    print("The keyword else is not in d. How this can be")
else:
    print("This keyword was found")
```

```python
l = d.items() # returns a list of tuples
```

```python
l
```

#### now we can iteration over them the same way we iterated over [[1,2],[2,2]]

```python
for k,v in d.items():
    print(f"{k} = {v}")
```
