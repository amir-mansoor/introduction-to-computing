```python
import os
directory = 'res'

filename = os.path.join(directory,"products-100.csv")
```

```python
print(filename)
```

```python
with open(filename, 'r') as f:
    for line in f:
        print(line.strip().split(','))
```

```python
rows = []

with open(filename, 'r') as f:
    for line in f:
        print(line.strip().split(','))
        rows.append(line.strip().split(','))
```

```python
rows
```

```python
for row in rows[1:]:
    for i in range(2):
        row[-1] = float(row[-1])
```

```python
rows
```

### Writing CSV Files

```python
','.join(['a','b','c'])
```

```python
for r in rows:
    print([str(c) for c in r])
```

```python
out_rows = []

for row in rows:
    r = [str(c) for c in row]
    out_rows.append(r)
```

```python
out_rows
```

```python
for r in out_rows:
    out_line = ','.join(r)
    print(out_line)
```

```python
csv_filename = os.path.join(directory, 'products-100-updated.csv')
```

```python
with open(csv_filename, 'w') as f:
    for r in out_rows:
        out_line = ','.join(r)
        f.write(out_line + '\n')
```
