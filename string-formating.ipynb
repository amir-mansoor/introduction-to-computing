```python
x = 24
y = 97
```

```python
print("x = ", x, "and y = ", y)
```

```python
print("x=%d and y=%d" % (x,y))
```

```python
print(f"x={x} and y={y}")
```

```python
x = 24.5887676
y = 97
```

```python
print("x=%f and y=%d" % (x,y))
```

```python
print("x=%.2f and y=%d" % (x,y))
```

```python
l = [1,98,23,198]
for i in l:
    print("%04d" % i) # or can use 0-paddin
```

```python
for i in l:
    print("{:3}".format(i))
```

```python
for i in l:
    print("{:03}".format(i))
```
