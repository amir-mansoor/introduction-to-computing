```python
try:
    print("Div 4 by 2 is", div(4,2))
    print("Div 4 by 0 is", div(4,0))
#     print("Hello world")
except ZeroDivisionError:
    print("Do not ask me to divide by 0 please.")
print("Done")
```

## Bubbling Exceptions

```python
def calculator(op,num1,num2):
    if op == 'div':
        return div(num1,num2)
    elif op == 'mul':
        return mul(num1,num2)
```

```python
print(calculator('div', 4,2))
```

```python
print(calculator('div', 4,0))
```

```python
try:
    print("Divide 4 by 0", calculator('div', 4, 0))
except ZeroDivisionError:
    print("Do not ask me to divide by zero please")
```

## Exception Hierarchy

```python
try:
    print("Divide 4 by 0", calculator('div', 4, 2))
    v
except ZeroDivisionError:
    print("Do not ask me to divide by zero please")
```

```python
try:
    print("Divide 4 by 0", calculator('div', 4, 2))
    v
except ZeroDivisionError:
    print("Do not ask me to divide by zero please")
except NameError:
    print("That's a bad name..")
```

```python
try:
    calculator('div', 4,0)
    v
except Exception:
    print("No idea what went wrong") # hence don't do this
```

## Raising Exceptions

```python
# remeber square root example
def sqrt(x, guess = 1.0):
    if x < 0:
        return None
    
#     calculate square root and return value

    return 6
```

```python
print(sqrt(-2))
```

```python
# remeber square root example
def sqrt(x, guess = 1.0):
    if x < 0:
        raise ValueError("Cannot find square root of negative numbers")
    
#     calculate square root and return value

    return 6
```

```python
sqrt(-2) # it's the caller responsibility to handle exception
```

```python
try:
    sqrt(-2)
except ValueError:
    print("Tried to find square root of negative numbers")
```
