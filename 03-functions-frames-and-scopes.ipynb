### Defining New Functions 

We can define the new functions using the def keyword 

```python
def square(x):
    return x*x  # Notice the indentation
```

And we *Call* this function as before

```python
square(5)
```

```python
v = square(5)
```

```python
v
```

```python
def sum_squares(x,y):
    return square(x) + square(y)
```

```python
result = sum_squares(5,5)
```

```python
result
```

### Obviously, we can call a function inside other functions as we did in sum_squares

```python
sum_squares("a",3)
```
