## Iteration,Repetition, Loops

### Sometimes you want to do things more than once 

```python
print("Put milk in")
print("Put sugar in")
print("put tea leaves in")
```

### However that's a lot of wastage of our time "A good programmer is a lazy programmer"

```python
for ingredient in 'milk', 'sugar', 'tea leaves':
    print(f"Put {ingredient} in.")
```

#### This is a compount statement that loops over the body. This particularly type, suprisingly, as termed as the for loop. It knows how many times we have to run the loop

```python
for i in 2,3,4,5,6,7,8,9:
    print(i)
    
print("Done")
```

```python
for i in range(2,10):
    print(i)
print("Done")
```

#### Sometimes we don't know how many times to run the loop...
... But we do know when to stop. This is called the while loop.
For example we want to print all multiples of 0.13 from 0 to 1.
But we don't know how many of those there are.

```python
n = 0.13
while n < 1:
    print(n)
    n = n + 0.13
    
# print(n)
print("Done")
```

```python
n = 0.13
i = 0 # our counter
while n < 1:
    i = i+1
    print(n)
    n = n + 0.13
    
print(i)
# print(n)
print("Done")
```

```python
for i in range(2,10):
    if i % 2 == 0:
        print(i)
```

```python
for i in range(2,10):
    if not i % 2 == 0:
        print(i)
```

```python
for i in range(2,10):
    if not i % 2 == 0:
        continue
            
    print(i)
```

```python
for i in range(2,10):
    if i % 5 == 0:
        break
            
    print(i)
```

### Factorial
##### Factorial is a useful function. it is used to calculate sin,cos and in several series as well as to calculate number of permutations etc. 
##### Factorial of a number is the multiplication of all integers up to that number. For example
##### fact(4) = 4x3x2x1 = 24


```python
fac = 1

for i in range(1,5):
    fac = fac * i
    
print(fac)
```

```python
fac = 1
i = 1
while(i <= 120):
    fac = fac * i
    i = i+1
print(fac)
```

### Fibonacci Numbers
Fibonacci Numbers are quite useful. Here's what they look like 1,1,2,3,5,8,13,21....
fib(0) = 0
fib(1) = 1
fib(n) = fib(n-1) + fib(n-2)
What good are those, you ask?

```python
def fib(n):
    if n <= 1:
        return n
    
    return fib(n-1) + fib(n-2)
    
fib(4)
```

```python
fib(10)
```

```python
%time fib(40)
```

```python
def fib2(n):
    if n <= 1: return n
    
    a,b = 0,1
    
    for i in range(1,n+1):
        n = a + b
        a = b
        b = n
        
    
    return a
```

```python
%time fib2(35)
```
